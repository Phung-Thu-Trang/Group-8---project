```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('../data/raw/Problem_Statement.xlsx')
print(df.head())
```

```python
# Ví dụ tạo dữ liệu mẫu:
data = {
    "Crime Type": ["Flora crimes", "Heroin trade", "Cyber-dependent crimes", "Extortion and protection racketeering", "Mafia-style groups","Government transparency and accountability","Victim and witness support"],
    "Index": [4.06,4.08,4.55,4.02,4.02,4.36,4.24]
}
df = pd.DataFrame(data)

# Vẽ biểu đồ thanh ngang
plt.figure(figsize=(10, 6))  # kích thước biểu đồ
plt.barh(df["Crime Type"], df["Index"], color=["#800026" if crime == "Cyber-dependent crimes" else "#F6DED8" 
                                                for crime in df["Crime Type"]])
 # barh = horizontal bar
plt.xlabel("Index")
plt.ylabel("Crime Type")
plt.title("Global crime type index in 2023")
plt.grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```
